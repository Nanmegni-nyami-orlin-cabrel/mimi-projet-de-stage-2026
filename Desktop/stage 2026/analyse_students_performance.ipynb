```python
import pandas as pd
import numpy as np

data=pd.read_csv("/StudentsPerformance.csv")
# Création de la moyenne générale
data['moyenne_generale'] = data[['math score', 'reading score', 'writing score']].mean(axis=1)

# 1. Meilleure moyenne
best = data.loc[data['moyenne_generale'].idxmax()]
print(f"Meilleure moyenne : {best['moyenne_generale']}")
# Interprétation : L'élève en tête possède un score parfait de 100 dans les trois matières.

# 2. Genre et Maths
print(data.groupby('gender')['math score'].mean())
# Interprétation : Les garçons obtiennent en moyenne un meilleur score en mathématiques que les filles.

# 3. Meilleur groupe en lecture
print(data.groupby('race/ethnicity')['reading score'].mean().sort_values(ascending=False))
# Interprétation : Le groupe E est le groupe ethnique qui réussit le mieux les tests de lecture.

# 4. Élèves en dessous de 50
echec = data[(data['math score'] < 50) | (data['reading score'] < 50) | (data['writing score'] < 50)]
print(f"Nombre d'élèves concernés : {len(echec)}")
# Interprétation : 145 élèves n'ont pas atteint la moyenne de 50 dans au moins une des trois matières.

# 5. Impact éducation parents
print(data.groupby('parental level of education')['moyenne_generale'].mean().sort_values(ascending=False))
# Interprétation : Plus le niveau d'étude des parents est élevé (Master/Bachelor), meilleurs sont les résultats des enfants.


display(data)
```
